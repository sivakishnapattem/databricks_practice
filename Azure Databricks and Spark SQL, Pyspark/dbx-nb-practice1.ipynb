```python
%sql

create catalog if not exists population_metrics;
use catalog population_metrics;
create volume if not exists population_metrics.default.population_data;
```

Once you create the catalog and the volumne, upload the CSV folder into the population_metrics.default.population_data volume manually

```python
path = '/Volumes/population_metrics/default/population_data/csv/countries_population.csv'
```

```python
df = spark.read.csv(path, header=True, inferSchema=True)
display(df)
```

```python
df = spark.read.load(path, format = 'csv', header = True, inferSchema = True)
display(df)
```

```python
schema1 = """
    country_id int,
    name string,
    nationality string,
    country_code string,
    iso_alpha2 string,
    capital string,
    population int,
    area_km2 double,
    region_id int,
    sub_region_id int"""
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType

schema2 = StructType([
    StructField('country_id', IntegerType(), False),
    StructField('name', StringType(), True),
    StructField('nationality', StringType(), True),
    StructField('country_code', StringType(), True),
    StructField('iso_alpha2', StringType(), True),
    StructField('capital', StringType(), True),
    StructField('population', IntegerType(), True),
    StructField('area_km2', DoubleType(), True),
    StructField('region_id', IntegerType(), True),
    StructField('sub_region_id', IntegerType(), True)
])
```

```python

df1 = spark.read.format('csv').options(header=True).schema(schema1).load(path)
display(df1)
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType


schema2 = StructType([
    StructField('country_id', IntegerType(), False),
    StructField('name', StringType(), True),
    StructField('nationality', StringType(), True),
    StructField('country_code', StringType(), True),
    StructField('iso_alpha2', StringType(), True),
    StructField('capital', StringType(), True),
    StructField('population', IntegerType(), True),
    StructField('area_km2', DoubleType(), True),
    StructField('region_id', IntegerType(), True),
    StructField('sub_region_id', IntegerType(), True)])

df4 = spark.read.format('json').schema(schema2).load(output_path)

df4.display()


```

```python
df4.write.mode('overwrite').format('parquet').save('/Volumes/population_metrics/default/population_data/output/parquet/countries_population')
df5 = spark.read.format('parquet').load('/Volumes/population_metrics/default/population_data/output/parquet/countries_population')
df5.display()
```

```python
df5.write.format('delta').save('/Volumes/population_metrics/default/population_data/output/delta/countries_population')
```
